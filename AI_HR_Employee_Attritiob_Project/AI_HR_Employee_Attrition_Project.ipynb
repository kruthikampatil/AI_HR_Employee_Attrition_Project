```python
import pandas as pd

df = pd.read_csv("HR_Employee_Attrition_500.csv") # Load Dataset

print(df.head()) # Show First 5 Rows

print(df.shape) # Dataset Shape
```

```python
import matplotlib.pyplot as plt

print(df['Attrition'].value_counts()) # Attrition Count

df['Attrition'].value_counts().plot( # Pie Chart
    kind='pie',
    autopct='%1.1f%%'
)

plt.title("Employee Attrition Distribution")
plt.ylabel("")
plt.show()
```

```python
import matplotlib.pyplot as plt

# Attrition Count
print(df['Attrition'].value_counts())

# Create Figure
plt.figure(figsize=(12,5))

# Chart 1
plt.subplot(1,2,1)

df['Attrition'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%'
)

plt.title("Employee Attrition Distribution")
plt.ylabel("")

# Chart 2
plt.subplot(1,2,2)

df['Gender'].value_counts().plot(
    kind='bar'
)

plt.title("Gender Distribution")

# Show All Charts
plt.tight_layout()
plt.show()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

le = LabelEncoder() # Convert text to numbers

df['Gender'] = le.fit_transform(df['Gender'])
df['Attrition'] = le.fit_transform(df['Attrition'])

X = df[['Age', 'MonthlyIncome', 'JobSatisfaction', 'YearsAtCompany', 'Gender']] # Features and Target

y = df['Attrition']

X_train, X_test, y_train, y_test = train_test_split( # Split Data
    X, y, test_size=0.2, random_state=42
)

model = DecisionTreeClassifier() # Split Data

model.fit(X_train, y_train) # Train Mod

y_pred = model.predict(X_test) # Prediction

accuracy = accuracy_score(y_test, y_pred) # Accuracy

print("AI Model Accuracy:", accuracy)
```

```python
new_employee = [[28, 35000, 2, 2, 1]] # New Employee Data

prediction = model.predict(new_employee) # Prediction

# Result
if prediction[0] == 1:
    print("Employee May Leave Company")
else:
    print("Employee Will Stay")
```
